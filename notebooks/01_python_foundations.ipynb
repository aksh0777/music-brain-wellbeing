## 1. Variables and objects
### Concept
A variable is a named reference (pointer) bound to an object stored in heap memory.
### Why it exists
To store and manipulate objects in computer memory using human-readable names without managing physical RAM addresses.
### First-principles intuition
A variable is a sticky name tag attached to a physical object sitting in a warehouse room (heap memory).
### Tiny example
Assigning participant ID `"P001"` to a variable `participant_id`.
### Code


````python
participant_id = "P001"
print("Participant ID:", participant_id)
````

### Expected output
`Participant ID: P001`
### Music/Data Science connection
Used across data cleaning scripts to reference participant IDs, raw file paths, and feature names.
### Common mistake
Believing `b = a` copies the underlying object instead of creating a second pointer to the exact same memory address.
### Interview question
Are Python variables values or pointers? *Answer: Pointers (references) to `PyObject` structures in heap memory.*


## 2. Integers and floats
### Concept
Numerical primitives representing discrete counts (`int`) and continuous decimal values (`float`).
### Why it exists
Data science calculations require both exact counting (e.g. sample size) and continuous measurement (e.g. signal voltage).
### First-principles intuition
An `int` counts individual items; a `float` measures continuous quantities along a real number line.
### Tiny example
Listening time in seconds (`320`) vs EEG alpha power in microvolts (`12.45`).
### Code


````python
duration_sec = 320
eeg_alpha_uv = 12.45
print(f"Duration: {duration_sec}s, Alpha Power: {eeg_alpha_uv} uV")
````

### Expected output
`Duration: 320s, Alpha Power: 12.45 uV`
### Music/Data Science connection
Representing song durations, sampling frequencies, and continuous biosignal amplitudes.
### Common mistake
Expecting floating-point arithmetic to be exact (e.g. `0.1 + 0.2 == 0.30000000000000004` due to IEEE 754 binary representation).
### Interview question
Why are integers in Python immutable? *Answer: Integers are immutable C structs; modifying an integer creates a new object in memory.*


## 3. Strings
### Concept
Immutable sequences of textual characters (`str`).
### Why it exists
To represent text, categorical labels, file paths, and track titles.
### First-principles intuition
A chain of character symbols locked in a fixed order that cannot be changed in place.
### Tiny example
Formatting genre tag `"Ambient"` to uppercase.
### Code


````python
genre = "Ambient"
print(f"Genre label: {genre.upper()}")
````

### Expected output
`Genre label: AMBIENT`
### Music/Data Science connection
Processing raw CSV header names, track titles, and categorical survey categories.
### Common mistake
Attempting string mutation like `s[0] = 'a'`, which raises a `TypeError`.
### Interview question
What is the difference between string concatenation (`+`) vs `.join()` in a loop? *Answer: `.join()` allocates memory once ($O(N)$), while `+` in a loop re-allocates memory repeatedly ($O(N^2)$).*


## 4. Boolean values
### Concept
Binary truth primitives (`True` or `False`).
### Why it exists
To direct control flow and evaluate filtering conditions.
### First-principles intuition
A light switch that is either ON (`True`) or OFF (`False`).
### Tiny example
Checking if stress score is high (>= 7).
### Code


````python
stress_score = 8
is_high_stress = stress_score >= 7
print("High Stress Flag:", is_high_stress)
````

### Expected output
`High Stress Flag: True`
### Music/Data Science connection
Flagging outlier recordings, filtering invalid participant rows, and creating binary target labels.
### Common mistake
Confusing the assignment operator `=` with the comparison operator `==`.
### Interview question
What are 'truthy' and 'falsy' values in Python? *Answer: Empty collections (`[]`, `{}`), `0`, `None`, and `""` evaluate to `False`; non-empty values evaluate to `True`.*


## 5. Lists
### Concept
Mutable, ordered sequence of object references (`[item1, item2]`).
### Why it exists
To store dynamic collections of data items that grow or shrink over time.
### First-principles intuition
A dynamic list of pointers pointing to various objects in memory.
### Tiny example
Appending daily music listening hours to a list.
### Code


````python
daily_hours = [1.5, 2.0, 0.5, 3.0]
daily_hours.append(2.5)
print("Daily Hours List:", daily_hours)
````

### Expected output
`Daily Hours List: [1.5, 2.0, 0.5, 3.0, 2.5]`
### Music/Data Science connection
Storing sequential features, participant IDs, or raw biosignal sample windows.
### Common mistake
Modifying a list while looping over it, which shifts indices and skips elements silently.
### Interview question
What is the time complexity of `list.append()` vs `list.insert(0, val)`? *Answer: `append()` is $O(1)$ amortized; `insert(0, val)` is $O(N)$ because every existing element must shift right.*


## 6. Dictionaries
### Concept
Mutable key-value mapping (`{key: value}`) backed by a Hash Table.
### Why it exists
To retrieve records instantly by name rather than searching through numeric index positions.
### First-principles intuition
A physical dictionary where word labels immediately point to their definitions.
### Tiny example
Storing track metadata and retrieving `"artist"` safely using `.get()`.
### Code


````python
track = {"title": "Weightless", "artist": "Marconi Union", "tempo": 60.0}
print("Artist:", track.get("artist", "Unknown"))
````

### Expected output
`Artist: Marconi Union`
### Music/Data Science connection
Representing survey log records, participant metadata, and model hyperparameter configs.
### Common mistake
Accessing missing keys using direct brackets `d[key]` which crashes with `KeyError` instead of using `d.get(key, default)`.
### Interview question
How does a Python `dict` achieve average $O(1)$ lookup time complexity? *Answer: It hashes the key using `hash(key)` to map directly to an array bucket index.*


## 7. Sets
### Concept
Unordered collection of unique, hashable items (`{item1, item2}`).
### Why it exists
To deduplicate values and perform set operations (intersections, unions) at $O(1)$ speed.
### First-principles intuition
A bag of distinct tags where duplicates automatically disappear.
### Tiny example
Extracting unique music genres from a list with duplicates.
### Code


````python
raw_genres = ["Ambient", "Classical", "Ambient", "Jazz", "Ambient"]
unique_genres = set(raw_genres)
print("Unique Genres:", unique_genres)
````

### Expected output
`Unique Genres: {'Classical', 'Ambient', 'Jazz'}`
### Music/Data Science connection
Finding distinct participants or unique genre tags across dataset files.
### Common mistake
Trying to add mutable items (like a `list`) into a `set`, which raises `TypeError: unhashable type: 'list'`.
### Interview question
What is the difference in membership testing (`x in container`) between a `list` and a `set`? *Answer: `list` is $O(N)$ linear search; `set` is $O(1)$ hash table lookup.*


## 8. Tuples
### Concept
Immutable, ordered sequence of elements (`(item1, item2)`).
### Why it exists
To create fixed, read-only data records that guarantee data integrity.
### First-principles intuition
A sealed container of items that can be read but never modified or resized.
### Tiny example
Storing fixed 3D EEG sensor coordinates `(x, y, z)`.
### Code


````python
sensor_location = (12.5, 45.0, 10.2)
print("Sensor Location tuple:", sensor_location)
````

### Expected output
`Sensor Location tuple: (12.5, 45.0, 10.2)`
### Music/Data Science connection
Storing electrode channel coordinates, matrix dimensions `(rows, cols)`, and fixed metadata tuples.
### Common mistake
Attempting to modify a tuple `t[0] = val`, which raises `TypeError: 'tuple' object does not support item assignment`.
### Interview question
Why are tuples faster and lighter than lists? *Answer: Tuples are stored in a single fixed memory block without extra allocation capacity overhead.*


## 9. Indexing
### Concept
Retrieving a specific element by its positional offset (`sequence[index]`).
### Why it exists
To access individual items in ordered sequences directly.
### First-principles intuition
Counting steps from the start (0, 1, 2) or steps backwards from the end (-1, -2).
### Tiny example
Accessing the first (`[0]`) and last (`[-1]`) biosignal sample.
### Code


````python
samples = [102.1, 104.5, 101.8, 105.0]
first_sample = samples[0]
last_sample = samples[-1]
print(f"First: {first_sample}, Last: {last_sample}")
````

### Expected output
`First: 102.1, Last: 105.0`
### Music/Data Science connection
Retrieving baseline signals at $t=0$ or selecting specific feature columns.
### Common mistake
Requesting an index equal to or greater than the length of the list, triggering `IndexError`.
### Interview question
Why does Python use zero-based indexing? *Answer: The index represents the offset distance in memory from the array start address.*


## 10. Slicing
### Concept
Extracting a sub-sequence range using `sequence[start:stop:step]`.
### Why it exists
To isolate specific windows of sequential data.
### First-principles intuition
Cutting a slice of tape starting at `start` (inclusive) up to `stop` (exclusive).
### Tiny example
Slicing a 4-element window `[2:6]` from a signal list.
### Code


````python
signal = [10, 20, 30, 40, 50, 60, 70, 80]
window = signal[2:6]
print("Signal Window [2:6]:", window)
````

### Expected output
`Signal Window [2:6]: [30, 40, 50, 60]`
### Music/Data Science connection
Windowing time-series biosignals (e.g. taking the last $N$ seconds of EEG recordings using `[-N:]`).
### Common mistake
Forgetting that `stop` index is exclusive (not included in the slice).
### Interview question
Does slicing a Python list create a new copy or a view? *Answer: Slicing a Python list creates a shallow copy of the sub-sequence.*


## 11. if/elif/else
### Concept
Branching logic to execute different code paths based on boolean evaluations.
### Why it exists
To enable dynamic decision making depending on data values.
### First-principles intuition
A fork in the road: if condition is `True`, take path A; elif next condition is `True`, take path B; else take path C.
### Tiny example
Categorizing a participant score into High, Moderate, or Low.
### Code


````python
wellbeing_score = 65
if wellbeing_score >= 75:
    category = "High"
elif wellbeing_score >= 50:
    category = "Moderate"
else:
    category = "Low"
print(f"Wellbeing Score {wellbeing_score} -> Category: {category}")
````

### Expected output
`Wellbeing Score 65 -> Category: Moderate`
### Music/Data Science connection
Segmenting users into stress tiers, validating data bounds, and handling pipeline flags.
### Common mistake
Writing redundant conditions like `if condition == True:` instead of `if condition:`.
### Interview question
How does Python evaluate complex `if` statements with `and` / `or`? *Answer: Via short-circuit evaluation—it stops evaluating as soon as the outcome is guaranteed.*


## 12. for loops
### Concept
Iterating directly over objects in a sequence (`for item in sequence:`).
### Why it exists
To execute repetitive code across every item in a collection.
### First-principles intuition
Stepping through a line of objects one-by-one until the end of the line.
### Tiny example
Summing track durations in seconds using direct iteration.
### Code


````python
durations = [300, 180, 240, 420]
total_sec = 0
for d in durations:
    total_sec += d
print("Total Duration (seconds):", total_sec)
````

### Expected output
`Total Duration (seconds): 1140`
### Music/Data Science connection
Iterating over dataset rows, computing aggregations, and running preprocessing pipelines.
### Common mistake
Using C-style index loops `range(len(arr))` when direct iteration `for item in arr:` is cleaner and safer.
### Interview question
How does a `for` loop work under the hood in Python? *Answer: It calls `iter(container)` to get an iterator and invokes `next()` until `StopIteration` exception is raised.*


## 13. while loops
### Concept
Repeated execution as long as a boolean condition remains `True` (`while condition:`).
### Why it exists
To repeat logic when the total number of iterations is unknown in advance.
### First-principles intuition
Keep running in a circle until someone signals stop.
### Tiny example
Processing a streaming signal buffer queue until empty (`buffer_size > 0`).
### Code


````python
buffer_size = 5
while buffer_size > 0:
    buffer_size -= 1
print("Buffer cleared! Final size:", buffer_size)
````

### Expected output
`Buffer cleared! Final size: 0`
### Music/Data Science connection
Streaming real-time biosignal buffers or running convergence loops in optimization algorithms.
### Common mistake
Forgetting to update the loop condition variable, causing an infinite loop.
### Interview question
When should you use a `while` loop instead of a `for` loop? *Answer: Use `while` when termination depends on a dynamic condition rather than iterating over a fixed sequence.*


## 14. Functions
### Concept
Named, reusable blocks of code (`def func(param):`) that encapsulate logic and return outputs.
### Why it exists
To eliminate code duplication, improve readability, and enforce modular design.
### First-principles intuition
A black-box machine: you feed inputs into the hopper, it processes them internally, and outputs a finished result.
### Tiny example
Converting seconds into minutes.
### Code


````python
def seconds_to_minutes(seconds):
    return seconds / 60.0

print("300 seconds =", seconds_to_minutes(300), "minutes")
````

### Expected output
`300 seconds = 5.0 minutes`
### Music/Data Science connection
Encapsulating feature extraction logic (e.g. converting raw voltages into spectral power densities).
### Common mistake
Forgetting the `return` statement, causing the function to implicitly return `None`.
### Interview question
What is the difference between positional and keyword arguments in Python? *Answer: Positional arguments depend on order; keyword arguments are passed by explicit parameter name.*


## 15. Basic exceptions
### Concept
Catching and handling runtime errors (`try: ... except ErrorType: ...`).
### Why it exists
To prevent pipeline crashes when encountering bad input data or missing files.
### First-principles intuition
Catching a falling glass before it shatters on the floor.
### Tiny example
Handling `ZeroDivisionError` safely in average calculation.
### Code


````python
def safe_divide(total, count):
    try:
        return total / count
    except ZeroDivisionError:
        return 0.0

print("Average with 0 count:", safe_divide(100, 0))
````

### Expected output
`Average with 0 count: 0.0`
### Music/Data Science connection
Handling missing survey keys, zero active listening duration, and missing raw files during ETL.
### Common mistake
Using bare `except:` which catches system exit signals and hides underlying bugs.
### Interview question
Why is catching specific exceptions preferred over bare `except:`? *Answer: Specific exceptions allow targeted error recovery without masking unexpected critical bugs.*


## 16. Imports and modules
### Concept
Importing external code libraries (`import math`) or custom modules (`from src.features import ...`).
### Why it exists
To organize code across multiple files and leverage specialized scientific libraries.
### First-principles intuition
Opening a reference book from your library shelf to use its formulas.
### Tiny example
Importing `math` module to compute square root.
### Code


````python
import math
print("Square root of 16:", math.sqrt(16))
````

### Expected output
`Square root of 16: 4.0`
### Music/Data Science connection
Importing internal processing modules (`src/features/`, `src/models/`) and scientific packages (`numpy`, `pandas`).
### Common mistake
Circular imports where Module A imports Module B while Module B imports Module A.
### Interview question
What happens under the hood when you execute `import module_name`? *Answer: Python searches `sys.path`, parses the code, executes it in a new module namespace, and caches it in `sys.modules`.*
