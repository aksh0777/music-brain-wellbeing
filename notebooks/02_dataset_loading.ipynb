## 0. Imports


```python
import pandas as pd
print("Pandas version:", pd.__version__)
```

## 1. pd.read_csv()

**What it does**: Reads a comma-separated values (CSV) file from disk into a Pandas DataFrame.

**Why it matters**: The raw data lives in a plain text file. `pd.read_csv()` parses every row and column, assigns inferred data types, and surfaces missing values as `NaN` automatically. This is always the first step when working with tabular data.


```python
# Load the raw dataset — do NOT modify the CSV file
CSV_PATH = "data/raw/mxmh_survey_results.csv"

df = pd.read_csv(CSV_PATH)
print("Dataset loaded successfully.")
```

## 2. df.shape

**What it does**: Returns a tuple `(number_of_rows, number_of_columns)`.

**Why it matters**: Immediately confirms the dataset scale — how many observations (rows) and how many variables (columns) exist.


```python
# How large is the dataset?
rows, cols = df.shape
print(f"Rows    : {rows}")
print(f"Columns : {cols}")
```

## 3. df.columns

**What it does**: Returns an Index object listing all column names.

**Why it matters**: Exact column names are required for every downstream selection, filter, and aggregation operation. Inspecting them first reveals naming conventions, spaces, or unexpected characters.


```python
# What columns exist?
print("Column names:")
for i, col in enumerate(df.columns):
    print(f"  [{i:>2}] {col}")
```

## 4. df.dtypes

**What it does**: Returns the inferred data type of every column — `int64`, `float64`, or `object` (text/string).

**Why it matters**: Confirms that numeric columns are stored as numbers (not text strings), and flags columns that may need type correction before analysis.

- `int64` — integer (whole number)  
- `float64` — floating point decimal  
- `object` — Python string / mixed types


```python
# What are the data types?
print("Column data types:")
print(df.dtypes)
```

## 5. df.head()

**What it does**: Returns the first 5 rows of the DataFrame.

**Why it matters**: Gives an immediate visual snapshot of what real data values look like — confirming formatting, range, and whether categorical columns contain expected labels.


```python
# Show the first 5 rows
print("First 5 rows:")
df.head()
```

## 6. df.info()

**What it does**: Prints a compact summary — total row count, column names, non-null counts per column, data types, and total memory usage.

**Why it matters**: Single most informative inspection call. It simultaneously shows missing value counts (rows with null values have lower non-null counts than the total) and type assignments. In Data Science interviews, `df.info()` is the first tool called after `pd.read_csv()`.


```python
# Full dataset health summary
print("Dataset info:")
df.info()
```

## 7. df.isnull().sum()

**What it does**:
- `df.isnull()` produces a boolean DataFrame — `True` where a value is missing, `False` where present.
- `.sum()` collapses each column to a single integer count of `True` values.

**Why it matters**: Reveals which columns have missing data, and how many rows are affected. This informs cleaning strategy choices BEFORE touching the data.


```python
# How many missing values per column?
missing = df.isnull().sum()
print("Missing value counts per column:")
print(missing)
print()
print(f"Total missing values across all columns: {missing.sum()}")
```
