## 1. Why NumPy exists
### Concept
NumPy solves performance bottlenecks and RAM inefficiencies when processing numerical datasets in standard Python.
### Why it exists
Python lists store scattered heap pointers, making mathematical calculations slow and memory-heavy.
### First-principles intuition
Upgrading from a chaotic pile of labeled envelopes in a room to a single organized egg carton.
### Tiny example
Comparing numerical storage efficiency.
### Code


```python
import numpy as np
print("NumPy module successfully loaded:", np.__version__)
```

### Expected output
`NumPy module successfully loaded: ...`
### What Python/NumPy is doing
Python executes code via dynamic bytecode interpretation; NumPy delegates numerical work to compiled C routines.
### Data Science connection
Stores raw biosignal readings and tabular feature metrics efficiently in hardware memory.
### Common mistake
Assuming standard Python lists can handle millions of math operations at real-time speeds.
### Interview question
What core problem does NumPy solve in Data Science? *Answer: It eliminates Python pointer/bytecode overhead by storing data in homogeneous contiguous C memory buffers.*


## 2. Python lists vs NumPy arrays
### Concept
Python lists store pointers to heterogeneous objects; NumPy arrays store homogeneous raw bytes in contiguous memory.
### Why it exists
To enable CPU cache optimization and vector hardware execution.
### First-principles intuition
A list of pointers vs a solid block of raw memory.
### Tiny example
Adding 10 to list items vs adding 10 to a NumPy array.
### Code


```python
py_list = [1, 2, 3]
py_res = [x + 10 for x in py_list]
np_arr = np.array([1, 2, 3])
np_res = np_arr + 10
print("List:", py_res, "| NumPy:", np_res)
```

### Expected output
`List: [11, 12, 13] | NumPy: [11 12 13]`
### What Python/NumPy is doing
Python loops evaluate type checks per element; NumPy passes contiguous C pointers directly to SIMD hardware instructions.
### Data Science connection
Allows mathematical operations to run 10x-100x faster across large datasets.
### Common mistake
Believing `list + list` adds numbers element-wise instead of concatenating list items.
### Interview question
How does a NumPy array differ in memory from a Python list? *Answer: A list is an array of heap pointers to scattered objects; an array is a single contiguous block of typed raw bytes.*

## 13. np.linspace()
### Concept
Generates a specified number of evenly spaced values over a specified interval inclusive.
### Why it exists
To sample an exact count of points across a continuous domain.
### First-principles intuition
Placing a fixed number of flags at equal intervals between start and end posts.
### Tiny example
Generating 5 points between 0 and 1 inclusive.
### Code


```python
l = np.linspace(0, 1, 5)
print("linspace(0, 1, 5):", l)
```

### Expected output
`linspace(0, 1, 5): [0.   0.25 0.5  0.75 1.  ]`
### What Python/NumPy is doing
Calculates exact division step $\frac{\text{stop} - \text{start}}{\text{num} - 1}$ and generates `num` values.
### Data Science connection
Generating sampling timestamps for continuous EEG wave plots.
### Common mistake
Confusing `np.arange` (takes step size) with `np.linspace` (takes total point count).
### Interview question
Difference between `np.arange` and `np.linspace`? *Answer: `arange` takes step size and excludes stop; `linspace` takes total point count and includes stop.*

## 18. reshape()
### Concept
Reinterprets array shape dimensions without altering data bytes.
### Why it exists
To change array layout for matrix operations without copying RAM.
### First-principles intuition
Re-arranging 8 eggs from an 2x4 carton into a 4x2 carton.
### Tiny example
Reshaping `np.arange(8)` to `(2, 4)`.
### Code


```python
arr = np.arange(8)
reshaped = arr.reshape(2, 4)
print("Reshaped (2, 4):\n", reshaped)
```

### Expected output
`Reshaped (2, 4):`
`[[0 1 2 3]`
` [4 5 6 7]]`
### What Python/NumPy is doing
Updates `shape` tuple metadata `(2, 4)` and stride values without moving underlying bytes in memory.
### Data Science connection
Structuring flat time-series vectors into 2D window matrices.
### Common mistake
Attempting to reshape to a shape whose size does not match total array elements.
### Interview question
Why does `reshape(2, 4)` work on an 8-element array? *Answer: $2 \times 4 = 8$ maintains total element size invariance; NumPy updates shape metadata in $O(1)$ time.*


## 19. reshape(-1, 1)
### Concept
Reshapes a 1D array into a 2D column vector with 1 column and automatically inferred row count.
### Why it exists
To format feature vectors into column matrices expected by machine learning algorithms.
### First-principles intuition
Turning a horizontal line of tiles into a single vertical column.
### Tiny example
Converting 1D vector to 2D column vector.
### Code


```python
v = np.array([10, 20, 30])
col = v.reshape(-1, 1)
print("Column Vector shape:", col.shape)
print("Column Vector:\n", col)
```

### Expected output
`Column Vector shape: (3, 1)`
`Column Vector:`
`[[10]`
` [20]`
` [30]]`
### What Python/NumPy is doing
Infers rows = $\frac{\text{total size}}{1} = 3$ and updates shape metadata to `(3, 1)`.
### Data Science connection
Passing single feature vectors to Scikit-Learn estimators expecting 2D input `(n_samples, 1)`.
### Common mistake
Believing `-1` creates a negative number instead of signaling automatic dimension inference.
### Interview question
What does `-1` mean in `reshape(-1, 1)`? *Answer: It tells NumPy to automatically calculate that dimension size based on total array size.*

```python
m = np.array([[1, 2, 3], [4, 5, 6]])
print("Sum axis=1:\n", np.sum(m, axis=1))
```

### Expected output
`Sum axis=1:`
`[ 6 15]`
### What Python/NumPy is doing
Iterates through memory strides along row offsets, summing elements across columns.
### Data Science connection
Calculating total listening time or mean stress score per individual participant (row).
### Common mistake
Confusing `axis=1` (collapses columns horizontally) with `axis=0` (collapses rows vertically).
### Interview question
What happens to array shape when reducing along `axis=1` on a `(2, 3)` matrix? *Answer: Column dimension 3 is collapsed, yielding output shape `(2,)`.*


## 30. vectorization
### Concept
Executing mathematical operations over whole arrays without explicit Python `for` loops.
### Why it exists
To achieve C-level speed and hardware optimization in numerical pipelines.
### First-principles intuition
Processing an entire assembly line at once instead of picking up items one-by-one by hand.
### Tiny example
Vectorized array multiplication benchmark.
### Code


```python
import time
a = np.arange(100000)
b = np.arange(100000)
t0 = time.perf_counter()
c = a * b
t1 = time.perf_counter()
print(f"Vectorized time: {(t1-t0)*1000:.2f} ms")
```

### Expected output
`Vectorized time: ... ms`
### What Python/NumPy is doing
Passes array pointers to compiled C code using SIMD CPU vector instructions.
### Data Science connection
Executing real-time signal transformations across millions of sensor samples.
### Common mistake
Writing manual Python `for` loops over NumPy arrays, destroying vectorization speed benefits.
### Interview question
Why is vectorization faster than Python loops? *Answer: It executes compiled C code over contiguous memory using SIMD hardware instructions, removing interpreter overhead.*


## 31. basic broadcasting
### Concept
Set of rules allowing element-wise operations between arrays of different shapes.
### Why it exists
To combine scalars or 1D vectors with 2D matrices without copying data memory.
### First-principles intuition
Automatically stretching a smaller array to fit a larger grid shape.
### Tiny example
Adding scalar to array and 1D vector to 2D matrix.
### Code


```python
m = np.array([[10, 20, 30], [40, 50, 60]])
v = np.array([1, 2, 3])
print("Matrix + Vector:\n", m + v)
```

### Expected output
`Matrix + Vector:`
`[[11 22 33]`
` [41 52 63]]`
### What Python/NumPy is doing
Aligns trailing shape dimensions; virtual strides replicate the smaller array without duplicating physical RAM bytes.
### Data Science connection
Normalizing feature columns by subtracting column mean vector $\mathbf{x} - \mathbf{\mu}$.
### Common mistake
Attempting to broadcast incompatible shapes where trailing dimensions do not match.
### Interview question
What condition allows a 1D vector of shape `(3,)` to broadcast with a 2D matrix of shape `(2, 3)`? *Answer: The trailing dimension matches (`3 == 3`), allowing the vector to broadcast across rows.*
