# Step 1 — Why Pandas Exists

### 1. What problem does Pandas solve?
Pandas provides high-level tabular data structures (`DataFrame` and `Series`) with integrated row/column labels, alignment, missing value handling, and database-style manipulation tools.

### 2. Why isn't Python enough?
Python lists and dictionaries lack columnar data alignment, fast vectorized operations across rows, built-in statistical summaries, and native CSV parsing.

### 3. Why isn't NumPy alone enough?
NumPy `ndarray` requires all elements to share the exact same data type (`homogeneous memory`). Real-world tabular data is **heterogeneous**: mixing text labels (`str`), ages (`int`), signal values (`float`), and dates (`datetime`) in different columns.

### 4. What is tabular data?
Tabular data is data organized into a 2D grid of **rows** (individual observations or participants) and **columns** (attributes or variables).

### 5. What is a DataFrame?
A `DataFrame` is a 2D, size-mutable, tabular data structure with labelled axes (rows and columns). Each column in a DataFrame is a `Series`.

### 6. What is a Series?
A `Series` is a 1D labelled array capable of holding any data type (`int`, `float`, `str`). It represents a single column or single variable in a DataFrame.

### 7. What is an Index?
An `Index` is an immutable sequence of labels associated with rows or columns that allows fast, explicit label-based lookup.

### Table Representation Example
Consider the simple table:

| Name | Age | Score |
| :--- | :--- | :--- |
| A | 20 | 80 |
| B | 22 | 91 |
| C | 21 | 75 |

Below is how Pandas represents this table in code.


```python
import pandas as pd
import numpy as np

# Representing the table in Pandas
data = {
    "Name": ["A", "B", "C"],
    "Age": [20, 22, 21],
    "Score": [80, 91, 75]
}

df_sample = pd.DataFrame(data)
print("Pandas DataFrame Representation:")
print(df_sample)
```

# Step 2 — Create and Inspect Data

We teach creating `Series` and `DataFrame`, followed by 8 essential inspection methods. Every inspection method explains **WHAT**, **WHY**, **WHAT IT RETURNS**, and **DATA SCIENCE RELEVANCE**.


```python
# Creating Series and DataFrame
s = pd.Series([10, 20, 30], name="Listening_Hours")
df = pd.DataFrame({
    "Participant": ["P1", "P2", "P3", "P4", "P5"],
    "Age": [24, 30, 21, 28, 35],
    "Genre": ["Ambient", "Classical", "Pop", "Rock", "Ambient"],
    "Score": [82.0, 75.0, 58.0, 64.0, 90.0]
})

print("Series:\n", s)
print("\nDataFrame:\n", df)
```

### df.head()
- **WHAT?**: Returns the first $N$ rows of the DataFrame (default 5).
- **WHY?**: To quickly verify data structure and column contents right after loading.
- **WHAT DOES IT RETURN?**: A DataFrame containing the top $N$ rows.
- **WHY IS IT USEFUL IN DATA SCIENCE?**: Allows immediate visual verification of column headers and data formatting.


### df.tail()
- **WHAT?**: Returns the last $N$ rows of the DataFrame (default 5).
- **WHY?**: To inspect footer rows, total count boundaries, or trailing anomalies.
- **WHAT DOES IT RETURN?**: A DataFrame containing the bottom $N$ rows.
- **WHY IS IT USEFUL IN DATA SCIENCE?**: Ensures file loading reached the true end of the dataset without truncation.


### df.shape
- **WHAT?**: Returns a tuple `(n_rows, n_cols)` representing dataset dimensions.
- **WHY?**: To quantify total observations (rows) and feature attributes (columns).
- **WHAT DOES IT RETURN?**: A 2-element integer tuple `(rows, columns)`.
- **WHY IS IT USEFUL IN DATA SCIENCE?**: Confirms expected sample counts and feature dimensionality.


### df.columns
- **WHAT?**: Returns the Index array of column header names.
- **WHY?**: To check exact feature names and detect leading/trailing whitespace issues.
- **WHAT DOES IT RETURN?**: An `Index` object containing column names.
- **WHY IS IT USEFUL IN DATA SCIENCE?**: Essential for referencing correct column keys in downstream transformations.


### df.index
- **WHAT?**: Returns the row Index label sequence.
- **WHY?**: To verify row indexing structure (range index or custom labels).
- **WHAT DOES IT RETURN?**: An `Index` object (e.g. `RangeIndex(start=0, stop=5)`).
- **WHY IS IT USEFUL IN DATA SCIENCE?**: Ensures row alignment is intact prior to joining or merging datasets.


### df.dtypes
- **WHAT?**: Returns the data type (`int64`, `float64`, `object`) for each column.
- **WHY?**: To verify that numbers are stored as numeric types and text as strings/objects.
- **WHAT DOES IT RETURN?**: A Series mapping column names to their data types.
- **WHY IS IT USEFUL IN DATA SCIENCE?**: Detects type bugs (e.g. numbers accidentally loaded as strings due to dirty formatting).


### df.info()
- **WHAT?**: Prints a concise summary including memory usage, non-null counts, and dtypes.
- **WHY?**: To audit total dataset health, missing value counts, and memory footprint in one view.
- **WHAT DOES IT RETURN?**: `None` (prints directly to output console).
- **WHY IS IT USEFUL IN DATA SCIENCE?**: First diagnostic tool used upon ingesting raw CSV data.


### df.describe()
- **WHAT?**: Generates summary statistics (mean, std, min, 25%, 50%, 75%, max) for numeric columns.
- **WHY?**: To inspect central tendency, dispersion, and detect extreme value outliers.
- **WHAT DOES IT RETURN?**: A DataFrame of summary metrics.
- **WHY IS IT USEFUL IN DATA SCIENCE?**: Identifies impossible values (e.g. negative age or 9999 placeholder values).

# Step 4 — Selecting Data

Accessing data columns and rows using direct brackets `df["col"]`, multiple brackets `df[["col1", "col2"]]`, label-based `.loc[]`, and positional `.iloc[]`.

### Key Distinction: `loc` vs `iloc`
- **`df.loc[row_label, col_label]`**: **Label-based selection**. Uses explicit row index labels and explicit column names. End slice index is **inclusive**.
- **`df.iloc[row_pos, col_pos]`**: **Positional integer selection**. Uses 0-based integer positional offsets. End slice index is **exclusive**.

### Output Prediction Prompts
Predict the selected series/dataframe output before executing each code block.


```python
# Setup DataFrame for selection
df_sel = pd.DataFrame({
    "Age": [24, 30, 21, 28],
    "Score": [82, 75, 58, 64],
    "Genre": ["Ambient", "Classical", "Pop", "Rock"]
}, index=["P1", "P2", "P3", "P4"])

print("Base DataFrame:\n", df_sel)

# 1. Single Column Selection (Returns Series)
# Predict: df_sel["Score"] -> ?
score_series = df_sel["Score"]
print("\nSingle Column ['Score'] (Series):\n", score_series)

# 2. Multiple Column Selection (Returns DataFrame)
# Predict: df_sel[["Genre", "Score"]] -> ?
multi_cols = df_sel[["Genre", "Score"]]
print("\nMultiple Columns [['Genre', 'Score']] (DataFrame):\n", multi_cols)

# 3. Label-based selection via loc
# Predict: df_sel.loc["P2", "Score"] -> ?
loc_single = df_sel.loc["P2", "Score"]
print("\nloc['P2', 'Score']:", loc_single)

# 4. Positional selection via iloc
# Predict: df_sel.iloc[1, 1] -> ?
iloc_single = df_sel.iloc[1, 1]
print("\niloc[1, 1]:", iloc_single)
```

# Step 5 — Filtering

Filtering selects specific DataFrame rows matching boolean conditions.

### Parentheses `()` Requirement for Multiple Conditions
In Python, bitwise operators `&` (AND) and `|` (OR) have **higher operator precedence** than comparison operators (`>`, `<`, `==`). Therefore, every condition MUST be enclosed in parentheses: `(df["col1"] > a) & (df["col2"] < b)`.

### `isin()`
The `.isin([val1, val2])` method filters rows where column values match any item in a specified list.


```python
df_filter = pd.DataFrame({
    "Participant": ["P1", "P2", "P3", "P4", "P5"],
    "Genre": ["Ambient", "Classical", "Pop", "Ambient", "Rock"],
    "Hours": [2.5, 1.8, 4.2, 5.0, 3.0],
    "Score": [82, 75, 58, 90, 64]
})

# Step 1: Single condition filtering
step1_high_hours = df_filter[df_filter["Hours"] > 3.0]
print("Step 1: Hours > 3.0:\n", step1_high_hours)

# Step 2: Multiple conditions using & (AND) with explicit parentheses
step2_multi = df_filter[(df_filter["Hours"] > 2.0) & (df_filter["Score"] >= 75)]
print("\nStep 2: Hours > 2.0 AND Score >= 75:\n", step2_multi)

# Step 3: Filtering using isin()
target_genres = ["Ambient", "Classical"]
step3_isin = df_filter[df_filter["Genre"].isin(target_genres)]
print("\nStep 3: Genre in ['Ambient', 'Classical']:\n", step3_isin)
```

# Step 6 — Data Cleaning

Handling missing data, duplicate records, and type conversions.

### Why Missing Data Exists
Data collection drops occurs due to unrecorded survey fields, hardware disconnects, or missing response entries.

### Trade-offs: Deleting vs Filling Missing Data
- **Blindly Deleting (`dropna()`)**: Dangerous because it discards entire valid observation rows, biasing samples and reducing statistical power.
- **Blindly Filling (`fillna()`)**: Dangerous because filling with arbitrary constants (like `0` or mean) distorts true distribution variance and introduces false noise.


```python
# Setup dataset with missing values and duplicates
df_dirty = pd.DataFrame({
    "ID": ["P1", "P2", "P3", "P1", "P4"],
    "Age": [24, np.nan, 21, 24, 35],
    "Score": [82.0, 75.0, np.nan, 82.0, 90.0]
})

print("Dirty DataFrame:\n", df_dirty)

# Step 1: Detect missing data
print("\nisna():\n", df_dirty.isna())
print("\nMissing count per column:\n", df_dirty.isna().sum())

# Step 2: Fill missing values safely
df_filled = df_dirty.copy()
df_filled["Age"] = df_filled["Age"].fillna(df_filled["Age"].mean())
print("\nAfter fillna on Age:\n", df_filled)

# Step 3: Detect and drop duplicates
print("\nduplicated():\n", df_dirty.duplicated())
df_clean = df_dirty.drop_duplicates()
print("\nAfter drop_duplicates():\n", df_clean)

# Step 4: Type conversion using astype()
df_clean_type = df_clean.copy()
df_clean_type["Score"] = df_clean_type["Score"].fillna(0).astype(int)
print("\nAfter astype(int):\n", df_clean_type.dtypes)
```

# Step 7 — Sorting and Basic Analysis

Analyzing categorical distributions and sorting rows.

- **`sort_values(by="col")`**: Sorts rows by values in specified column.
- **`value_counts()`**: Counts frequencies of unique category occurrences.
- **`unique()`**: Returns an array of distinct unique values.
- **`nunique()`**: Returns the total integer count of unique distinct values.


```python
df_analysis = pd.DataFrame({
    "Participant": ["P1", "P2", "P3", "P4", "P5"],
    "Genre": ["Ambient", "Classical", "Ambient", "Pop", "Ambient"],
    "Score": [82, 75, 90, 58, 77]
})

# 1. Sorting by score descending
step1_sorted = df_analysis.sort_values(by="Score", ascending=False)
print("Step 1: Sorted by Score (descending):\n", step1_sorted)

# 2. Category value counts
step2_counts = df_analysis["Genre"].value_counts()
print("\nStep 2: Genre value_counts():\n", step2_counts)

# 3. Unique and nunique
print("\nUnique Genres:", df_analysis["Genre"].unique())
print("Number of Unique Genres:", df_analysis["Genre"].nunique())
```

# Step 8 — GroupBy

The `groupby()` operation implements the **Split $\to$ Apply $\to$ Combine** pipeline:
1. **Split**: Divides the DataFrame into separate group buckets based on unique key values.
2. **Apply**: Computes an aggregation function (`mean()`, `sum()`, `count()`, `min()`, `max()`) independently on each group bucket.
3. **Combine**: Merges the aggregated results back into a single summary DataFrame.


```python
df_grp = pd.DataFrame({
    "Genre": ["Ambient", "Classical", "Ambient", "Classical", "Ambient"],
    "Score": [82, 75, 90, 70, 88],
    "Hours": [2.5, 1.8, 5.0, 2.0, 4.0]
})

# Step 1: Split and Apply mean aggregation
grp_mean = df_grp.groupby("Genre").mean()
print("Groupby Genre Mean:\n", grp_mean)

# Step 2: Split and Apply multiple aggregations explicitly
grp_stats = df_grp.groupby("Genre")["Score"].agg(["count", "mean", "min", "max"])
print("\nGroupby Genre Score Aggregations:\n", grp_stats)
```

# Step 9 — Merging

`pd.merge()` combines two DataFrames horizontally based on matching key columns (relational database joins).

### Join Types:
- **Inner Join** (default): Retains only rows with matching key values in BOTH tables.
- **Left Join**: Retains ALL rows from the left table and matching rows from the right table.
- **Right Join**: Retains ALL rows from the right table and matching rows from the left table.


```python
df_participants = pd.DataFrame({
    "ID": ["P1", "P2", "P3", "P4"],
    "Name": ["Alice", "Bob", "Charlie", "David"]
})

df_scores = pd.DataFrame({
    "ID": ["P1", "P2", "P3", "P5"],
    "Wellbeing": [82, 75, 90, 60]
})

# 1. Inner Join
inner_merge = pd.merge(df_participants, df_scores, on="ID", how="inner")
print("Inner Merge (matching keys only):\n", inner_merge)

# 2. Left Join
left_merge = pd.merge(df_participants, df_scores, on="ID", how="left")
print("\nLeft Merge (all left rows preserved):\n", left_merge)
```

# Step 10 — Concat

`pd.concat()` stacks DataFrames vertically along rows (`axis=0`) or glues them horizontally along columns (`axis=1`).

### Distinction: `concat` vs `merge`
- **`concat`**: Physical stacking/gluing of array blocks based on index positions or column names.
- **`merge`**: Relational database lookup based on matching key column values.


```python
df_batch1 = pd.DataFrame({"ID": ["P1", "P2"], "Score": [82, 75]})
df_batch2 = pd.DataFrame({"ID": ["P3", "P4"], "Score": [90, 64]})

# Stacking rows vertically (axis=0)
df_stacked = pd.concat([df_batch1, df_batch2], axis=0, ignore_index=True)
print("Vertical Concat (stacked rows):\n", df_stacked)
```
